```python
def odd(number_list):
    for i in number_list:
        if i % 2 != 0:
            return False
        else:
            pass
    return True
odd([8,6,2,4,2])
```

```python
def even_number_list(number_list):
    even_number = [number for number in number_list if number%2==0]
    return even_number 
even_number_list([23,1,3,5,7,9])
```

```python
workers = [('sam',800),('athi',300),('joshua',500),('jospher',600)]
```

```python
def employee_check(workers):
    current_max = 0
    employee_name =''

    for n,h in workers:
        if h > current_max:
            current_max = h
            employee_name = n
        else:
            pass
    return(employee_name,current_max)

         
```

```python
a,b =employee_check(workers)
print(a)
print(b)
```
