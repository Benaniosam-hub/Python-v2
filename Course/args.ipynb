```python
def myfunc(*sam):
    return sum(sam) * 0.05
myfunc(40,60,100,100)
```

```python
def functions(**kwargs):
    print(kwargs)
    if 'fruit' in kwargs:
        print(f'my fav fruit is {kwargs['fruit']}')
    else:
        print('I did not find any fruit here')

functions(fruit='Mango',veggies='letties')
```

```python
def arguments(*args,**kwargs):
    print('I want {} {} from the shop'.format(args[0],kwargs['fruit']))

arguments(10,20,30,veggies='letties',fruit='mangos')
```

```python
def squared(num):
    return num ** 2
```

```python
my_list=[1,2,3,4,5,6,7,8]
```

```python
for i in map(squared,my_list):
    print(i,",", end ="")
```

```python
list(map(squared,my_list))
```

```python
def naming(mystring):
    if len(mystring)%2 ==0:
        return 'EVEN'
    else:
        return mystring[0]
    
```

```python
mystring=['Benaniosam','Bejosam','Joshua','Jersha','Vijinsha']
```

```python
for item in map(naming,mystring):
    print(item)
```

```python
list(map(naming,mystring))
```
