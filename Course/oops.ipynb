```python
class Circle():

    pi = 3.14

    def __init__(self,radius=1):

        self.radius = radius
        self.area = radius*radius*Circle.pi 

    def get_circumference(self):
        return self.radius * Circle.pi * 2
```

```python
my_circle = Circle(30)
```

```python
my_circle.pi
```

```python
my_circle.area
```

```python
my_circle.get_circumference()
```
