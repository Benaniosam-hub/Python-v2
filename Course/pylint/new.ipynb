```python
def hello(name = 'sam'):
    print("Function has been executed")

    def greet():
        return "\t hello inside function greets you"
    def welcome():
        return "\t hello inside function welcomes you"

    if name == 'sam':
        return greet
    else:
        return welcome
```

```python
rev = hello('ram')
```

```python
rev()
```

```python
def new_decorator(original_function):
    def wrap_func():
        print("Wrapped up ")
    
        original_function()

        print("Some extra code")

    return wrap_func
```

```python
@new_decorator
def function():
    print("Hello there")
```

```python
function()
```

```python
def arigato(real_func):
    def wrapper():
        print("Konichiwa")
        real_func()
        print("Some Arigato")
    return wrapper
    
```

```python
@arigato
def new_func():
    print("New Funko Here!")
```

```python
new_func()
```

```python
def cubic(n):
    result = []
    for x in range (n):
        result.append(x ** 3)
    return result


```

```python
for x in cubic(11):
    print (x)
```

```python
def cubi(n):
    for x in range(n):
        yield x ** 3
```

```python
cubi(11)
```

```python
for i in cubi(11):
    print(i)
```

```python
def fibonacci_series(n):
    a=1
    b=1
    for x in range(n):
        yield a
        a,b = b,a+b
```

```python
for i in fibonacci_series(4):
    print(i)
```
