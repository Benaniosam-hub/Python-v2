```python
class Animal():
    def __init__(self):
        print("Animal created")
    def who_am_i(self):
        print("I am an animal")
    def eat(self):
        print("Im eating")
```

```python
class Dog(Animal):
    def __init__(self):
        Animal.__init__(self)
        print("Dog is created")
    def who_am_i(self):
        print("I am dog")
```

```python
class Cat(Animal):
    def __init__(self):
        Animal.__init__(self)
        print("Cat is created")
    def who_am_i(self):
        print("I am cat")
```

```python
class Book():
    def __init__(self,author,title,pages):
        self.title = title
        self.author = author
        self.pages = pages
    def __str__(self):
        return f"{self.title} by {self.author}"
    def __len__(self):
        return self.pages
    def __del__(self):
        print("A book object has been deleted")
```

```python
a = Book(title='Harrypotter',author='J.K.Rowling',pages=500)
```

```python
print(a)
```

```python
str(a)
```

```python
len (a)
```

```python
del a
```

```python
print(a)
```

```python
class Line:
    def __init__(self,coor1,coor2):
        self.coor1 = coor1
        self.coor2 = coor2
        
    def distance(self):
        x1,y1 = self.coor1
        x2,y2 = self.coor2
        d = ((x2 - x1)**2 + (y2 - y1)**2) ** 0.5
        return d
    def slope(self):
        x1,y1 = self.coor1
        x2,y2 = self.coor2
        m = (y2 - y1)/(x2 - x1)
        return m
```

```python
l =Line(coor1=(3,2),coor2=(8,10))
```

```python
l.distance()
```

```python
l.slope()
```
