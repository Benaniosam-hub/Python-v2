```python
sentence = "How you doing martha, long time no see, cause of what you did to me is well know for me."
```

```python
s = Counter(sentence.lower())
s.pop(' ')
s.pop('.')
s.pop(',')
```

```python
for i,w in s.items():
    print(i,'-',w)
```
